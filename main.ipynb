```python
import numpy as np
import pandas as pd
```

```python
students_dataset = np.array([[85,90,78,88],[72,65,80,70],[95,92,94,96],[60,75,68,72],[78,82,85,80]])
student_names = np.array(["Alice","Bob","Charlie","Diana","Ethan"])
students_sub = np.array(["Math","Python","Numpy","Stats"])
```

##Feature 1 — Create and inspect the dataset

```python
students_dim = np.ndim(students_dataset)
print("students dimension: ",students_dim)
students_shape = np.shape(students_dataset)
print("students shape: ",students_shape)
students_size = np.size(students_dataset)
print("students size:",students_size)
data_type = students_dataset.dtype
print("students data type",data_type)


##Access one student's marks and one subject's marks using indexing and slicing.

student_name_charlie_position = np.where(student_names == "Charlie")
sub_math = np.where(students_sub=="Math")
student_charlie_math_number = students_dataset[student_name_charlie_position[0][0],sub_math[0][0]]
charlie_numbers = students_dataset[student_name_charlie_position[0][0]]
print(f"Charlie's position {student_name_charlie_position[0][0]}: All marks {charlie_numbers}")
print(f"Math position {sub_math[0][0]}: Number {student_charlie_math_number}")

sub_numpy = np.where(students_sub=="Numpy")
numpy_marks = students_dataset[:,sub_numpy[0][0]]
print("Numpy marks: ",numpy_marks)

```

##Feature 2 — Calculate student statistics

```python
#Calculate each student's total marks.
each_students_total_mark = students_dataset.sum(axis=1)
print(each_students_total_mark)

each_students_average = students_dataset.mean(axis=1)
# each_students_average

#highest and lowest mark in the entire dataset
highest_mark = students_dataset.max()
highest_mark

lowest_mark = students_dataset.min()
lowest_mark

#the average mark for each subject
each_sub_avg_mark = students_dataset.mean(axis=0)
each_sub_avg_mark
```

```python
#new dataset table using pandas
new_students_data = pd.DataFrame(students_dataset,index=student_names,columns=students_sub)
new_students_data["Total mark"] = each_students_total_mark
new_students_data["Average"] = each_students_average
new_students_data
```
